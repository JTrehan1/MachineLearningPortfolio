<h1> Prinicple Component Analysis</h1>

It is a technique where we can reduce the number of features in a dataset without losing any information about the data. 

Variance is key to the amount of information a dataset provides.

Consider a dataset, some columns do no provide as much information as others. Example a column representing the sum of subsets across each row of data does not provide much information. This is due to the lack of variance across each entry in this row.

Along with variance the coefficient of variance which accounts for central tendencies of a dataset. Consider a dataset with variance of 5 centred around mean 5 and 100 - the interpretation is very different. 

**Principle components are the linear combination of all input features from the input dataset**


$$ CV = \frac{s}{\bar{x}} \times 100 $$

s = sample standard deviation 

$ \bar{x} = \text{sample mean} $

Taking the information and using it to reduce the dimensionality of the dataset. 

 -  Obtain a data matrix 
 - Calculate a covariance matrix 

 $$Cov(X,Y) = E[(X - E[X])(Y - E[Y])] $$

- Matrix factorisation, eigenvalues and eigenvectors

<h2>Matrix Factorisation</h2>
Find two matrices who's product is the covariance matrix. This allows us to have a matirx which captures the majority of the information needed.

<h2>Eigenvalues and Eigenvectors</h2>
Do not change direction when a transformation applied to them. In data context - provides a direction to rotate the dataset so it can be looked at from a simplified perspective. 

For a matrix A, the eigenvectors and eigenvalues are the solution to the following, known as the characterisic equation. 
$\det(A - \lambda I)$

  1) eigenvectors, also known as Principal Components, which define the direction, or "rotation", of our new data space
  2) eigenvalues, which determine the magnitude of that new data space


PCA is an unseupervised algorithm in its own nature. It identifies similarities in data to create clusters and used as a pre-processing step for clustering algorithms.

Especially useful in image classification.

Implementation: scikit-learn provides a more in-depth set of methods and attributes that extend the number of ways to perform PCA or display the percentage of variance for each principal axis. For example <code>.components_</code> and <code>pca.explained_variance_ratio_</code>


Dimensionality reduction: We visualised the data projected onto the principal axes, known as principal components.

Image classification: Performing PCA on images of faces to visually understand how principal components still retain nearly all the information in the original dataset.

Improved algorithmic speed/accuracy: Using principal components as input to a classifier, we observed how we could achieve equal or better results with lower dimensional data. Having lower dimensionality also speeds the training.
