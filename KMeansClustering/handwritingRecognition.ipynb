```python
import numpy as np
from matplotlib import pyplot as plt
from sklearn import datasets 
from sklearn.cluster import KMeans

digits = datasets.load_digits()
print(digits.DESCR)
print(digits.data)
print(digits.target)

# Image at index 100
plt.gray()
plt.matshow(digits.images[100])
plt.show()
# Label for the image @ index 100
print("Image 100 label:", digits.target[100])

# Size of the images are 8x8 so the datapoints (pixels) live in 64 dimensions

# Assuming decimal digits there should be 10 clusters, thus k = 10 

model = KMeans(n_clusters=10, random_state=42)
model.fit(digits.data)

# Visualise the centroid 
fig = plt.figure(figsize=(8,3))
plt.suptitle("Centroids for Image Data")

# Display each of the cluster centers
for i in range(10):
  # Plots in a 2x5 grid and iterate across
  ax = fig.add_subplot(2, 5, 1 + i)
  ax.imshow(model.cluster_centers_[i].reshape((8,8)), cmap=plt.cm.binary)

plt.show()
plt.clf()
```
