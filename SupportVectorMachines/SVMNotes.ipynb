<h1>Support Vector Machines</h1>

Supervised machine learning model used for classification. An SVm makes decisions by setting a decision boundary and seeing what side of the boundary an unclassified point falls on. These boundaries are defined by using a training set of classified points.

It is easiest to understand a decision boundary when the data has two features and the boundary is a line. If there are three features the decision boundary is a plane.

For data with more then three features it becomes more difficult to visualise the decision boundary, the boundary is then known as a separating hyperplane.


<h2>Optimal Decision Boundaries</h2>

One problem to be solved is figuring out what decision boundary to use. The optimal decision boundary needs to be as far away from the training points as possible. This decreases the chance of false classifications

Support vectors - points in the training set closest to the decision boundary and define the decision boundary. Consider them as vectors that originate at the origin. If there are n features there are at least n+1 support vectors. 

The distance between the support vector and the decision boundary is called the margin. The margin should be maximised. 

Since the support vectors are the most critical data point many other training points can be ignored and this makes SVMs more efficient.
