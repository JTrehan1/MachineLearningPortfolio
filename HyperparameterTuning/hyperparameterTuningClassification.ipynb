### 1. Explore the Dataset

```python
# 1. Setup
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform

raisins = pd.read_csv('Raisin_Dataset.csv')
raisins.head()
```

```python
# 2. Create predictor and target variables, X and y
y = raisins['Class']
X = raisins.drop(columns=['Class'])
```

```python
# 3. Examine the dataset
print(X.columns)
print(raisins.info())
print(raisins.describe())
```

```python
# 4. Split the data set into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 19)
```

### 2. Grid Search with Decision Tree Classifier

```python
# 5. Create a Decision Tree model
tree = DecisionTreeClassifier()
```

```python
# 6. Dictionary of parameters for GridSearchCV
parameters = {'max_depth': [3,5,7], 'min_samples_split': [2,3,4]}
```

```python
# 7. Create a GridSearchCV model
grid = GridSearchCV(tree, param_grid = parameters)

# Fit the GridSearchCV model to the training data
grid.fit(X_train, y_train)
```

```python
# 8. Print the model and hyperparameters obtained by GridSearchCV
best_model = grid.best_estimator_
print(best_model)

# Print best score
print(grid.best_score_)

# Print the accuracy of the final model on the test data
best_score_test = best_model.score(X_test, y_test)
print(best_score_test)
```

```python
# 9. Print a table summarizing the results of GridSearchCV
hyperparameter_value = pd.DataFrame(grid.cv_results_['params'])
print(hyperparameter_value)

mean_test_scores = pd.DataFrame(grid.cv_results_['mean_test_score'])
print(mean_test_scores)

df = pd.concat([hyperparameter_value, mean_test_scores], axis=1)
df.rename(columns={0:'Mean Test Score'}, inplace=True)
print(df)
```

### 2. Random Search with Logistic Regression

```python
# 10. The logistic regression model
lr = LogisticRegression(solver='liblinear', max_iter=1000)
```

```python
# 11. Define distributions to choose hyperparameters from
distributions = {'penalty':['l1', 'l2'], 'C':uniform(loc=0, scale=100)}
```

```python
# 12. Create a RandomizedSearchCV model
clf = RandomizedSearchCV(lr, distributions, n_iter = 8)

# Fit the random search model
clf.fit(X_train, y_train)
```

```python
# 13. Print best esimator and best score
best_model_2 = clf.best_estimator_
print(best_model_2)

best_score_2 = clf.best_score_
print(best_score_2)

# Print the accuracy of the final model on the test data
best_score_test_2 = best_model_2.score(X_test, y_test)
print(best_score_test_2)

# Print a table summarizing the results of RandomSearchCV
hyperparameter_value_2 = pd.DataFrame(clf.cv_results_['params'])
print(hyperparameter_value_2)

mean_test_scores_2 = pd.DataFrame(clf.cv_results_['mean_test_score'])
print(mean_test_scores_2)

df = pd.concat([hyperparameter_value_2, mean_test_scores_2], axis=1)
df.rename(columns={0:'Mean Test Score'}, inplace=True)
print(df)
```
