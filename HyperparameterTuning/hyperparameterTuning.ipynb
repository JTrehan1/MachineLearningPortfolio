<h1>Hyperparameters in Machine Leaning Models</h1>

***Hyperparameter*** - value that determines part of the learning process and is not affected by training. Common examples of hyperparameters include K in KNN and alpha in L1 and L2 regularisation methods. 

Hyperparameters allow for challenges such as Bias Variance Tradeoff to be addressed.

 - Bias is the difference between the model predictions and the correct values. Models with a lot of bias underfit the data - this makes them perform badly on test and training data.
 - Variance is the dependence of the model on training data. A model has high variance if different training data results in widely varying outcomes. This leads to overfitting the training data but poor model performance on the testing data. 

 Decreasing Bias -> Increases Variance 

<h1>Common Hyperparameter Tuning Methods</h1>

 - Grid search; works by training a model a predetermined list of hyperparameter values. This model tries every hyperparameter in the list, then uses the one that makes the model perform best.  
 - Random search; works similarly but on random values for the hyperparameter values. It selects the hyperparameter that performs best.
 - Bayesian optimisation; using bayesian techniques, it iterates through different hyperparameter values, and uses it to inform where to look next for the best hyperparameter.
 - Genetic algorithms; go through several generations of hyperparameters. Within each generation, the fittest is slightly mutated to produce the next generation of hyperparameter values. 

***Ensure data is split into training, test and validation sets when doing hyperparameter tuning***
 - Validation data is used to tune hyperparameters.
