<h1>Stacking</h1>

Stacking operates differently to bagging and boosting:
 
 The base estimators do not need to be weak learners.

 Unlike bagging there is no subsampling processes used. The stacking model effectively uses the complete training set.

It is a proven method for improving prediction accuracy of standalone models 

<h1>How Does Stacking Work?</h1>

Consider a classification problem: Using a decision tree we find the model is overfit to the training data and has poor generalisation. If a logistic regression model is used the model parameters are hard to tune because features are highly correlated. Each model has their own strengths and weaknesses - stacking helps balance this.

Stacking is where different models are trained subsequently and cast their vote through their predictions. A majority rules approach can be used for determining the final model prediction if each estimator is weighted equally.

Practically models may be weighted differently and as such the aggregation of base estimators will need to account for this.

<h1>Training Base Estimators</h1>
Different combinations of different base estimators can be used, for example logistic regression model and decision trees. Alternatively, the same learning algorithm can be used but with different model hyperparameters.

Since multiple models are trained on the same training data - overfitting becomes a high risk so K-fold Cross Validation methods are employed. 

Consider 10 segments and a stacking model that uses logistic regression and a decision tree model. Each model can be trained using data from 9 of the segments and make predictions on the excluded 10th segment. 

The predictions are then appended to the 10th segment, this means 1/10th of the data has two new features. One new feature is the logistic regression prediction and the other new feature is the decision tree prediction.

This is then done for all other 9 segments such that the processes is repeated for all segments of the training data. The result is a prediction made on each training sample without having seen the sample during the training process. 

<h1>Feature Augmentation</h1>
In our stacking setup, the base estimators need to be trained to make predictions on our training data. The prediction of each estimator will be appended to the corresponding data sample as a new feature. We thus augment the training data set with this additional information. The augmented training set is used by our later-stage stacking model to make the final prediction.

So, in summary, say our training dataset has 10,000 samples, 10 features, and we select 3 base estimators. We would train each base estimator on the training set and make predictions on the training set. Each estimator would make a prediction on each sample, therefore each sample will have 3 predictions. These 3 predictions are appended to the pre-existing 10 features. This leaves us with 10,000 training samples with 13 features each.

<h1>Training the stacking model</h1>
With the augmented training set, the final model can be prepared that will make the official prediction from the stacking setup.

This later stage model is selected just like a base estimator, an algorithm used earlier can also be reused. The purpose of this model is to learn the proper weighting of the earlier estimator given our training data now include a data point from each estimator. Some estimators will perform better than others so should be given a better weighting. 

The only difference in the training process is that the model will be designed to accept samples of the augmented size rather than the given size. This means if the given data set has n features and m base estimators, this model will require n + m features on each sample.

Once the later-stage model is trained, we feed testing data samples into the base estimators, which will append their predictions to the data sample. This sample is then given to the later-stage model for our final prediction.
